# LumiBench no Lumi Lab

Verificação das trilhas de conhecimento e decisão com quatro itens sintéticos e respostas fixas. A CLI também aceita um servidor OpenAI-compatible e o snapshot canônico do LumiBench.

Execute com o Python da sua sessão de Jupyter a partir do repositório. As duas etapas abaixo usam o mesmo código do terminal.


```python
from pathlib import Path
from tempfile import TemporaryDirectory
import json
import subprocess
import sys

project = Path.cwd().resolve()
if project.name == "notebooks":
    project = project.parent
assert (project / "benchmark/__main__.py").is_file(), "Abra o notebook dentro do Lumi Lab"

def benchmark(*args):
    return subprocess.run(
        [sys.executable, "-m", "benchmark", *map(str, args)],
        cwd=project, check=True, capture_output=True, text=True,
    )

print(f"Python {sys.version.split()[0]}")
```

```python
with TemporaryDirectory() as temp:
    run_dir, replay_dir = Path(temp) / "demo", Path(temp) / "replay"
    benchmark("run", "--suite", "demo", "--out", run_dir)
    benchmark("score", "--run", run_dir, "--out", replay_dir)
    original = (run_dir / "metricas.json").read_bytes()
    assert original == (replay_dir / "metricas.json").read_bytes()
    metrics = json.loads(original)

print("Conhecimento", metrics["conhecimento"]["geral"]["n"], "itens")
print("Decisão", metrics["decisao"]["geral"]["n"], "itens")
print("Recálculo offline idêntico")
```

Para usar o recorte real e trocar o backend, siga o [guia do benchmark](../benchmark/README.md). Use desenvolvimento para ajustar a configuração e preserve o teste para a medição final.
